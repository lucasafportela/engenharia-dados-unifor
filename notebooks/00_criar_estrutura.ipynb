```python
%sql
-- CATÁLOGO PRINCIPAL
CREATE CATALOG IF NOT EXISTS unifor;

    -- CAMADA LANDING
    -- Cria o schema que armazenará os arquivos brutos recebidos
    CREATE SCHEMA IF NOT EXISTS unifor.landing;
        -- Arquivos CSV da base pública da Olist
        CREATE VOLUME IF NOT EXISTS unifor.landing.olist;
        -- Arquivos JSON recebidos da API de usuários
        CREATE VOLUME IF NOT EXISTS unifor.landing.mock_users;
        -- Arquivos extraídos de banco PostgreSQL
        CREATE VOLUME IF NOT EXISTS unifor.landing.postgres;
        -- Arquivos do conjunto de dados Yellow Taxi
        CREATE VOLUME IF NOT EXISTS unifor.landing.yellow_taxi;

    -- ARQUITETURA MEDALHÃO
    -- Bronze: dados brutos estruturados em tabelas Delta
    CREATE SCHEMA IF NOT EXISTS unifor.bronze;
    -- Silver: dados limpos, padronizados e tratados
    CREATE SCHEMA IF NOT EXISTS unifor.silver;
    -- Gold: indicadores, agregações e dados prontos para análise
    CREATE SCHEMA IF NOT EXISTS unifor.gold;
```
